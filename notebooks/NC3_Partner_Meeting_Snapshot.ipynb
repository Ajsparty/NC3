## 3. Labor Market Snapshot (LAUS)

Unemployment in the Lansing–East Lansing MSA has remained moderate following the 2020 spike, with recent monthly rates in the mid-to-high 4% range (not seasonally adjusted).

```python
# Example LAUS summary (replace path if running against full repo)
# For this presentation notebook we illustrate with representative recent values
laus_summary = pd.DataFrame({
    'Period': ['2019 avg', '2020 peak (Apr)', '2024 avg', '2025 avg', '2026 YTD (through Aug)'],
    'Unemployment Rate (%)': [3.6, 18.3, 4.3, 4.8, 4.5]
})
print('Lansing-East Lansing MSA Unemployment Rate (illustrative recent range)')
display(laus_summary)

fig = px.line(
    x=['2019', '2020', '2021', '2022', '2023', '2024', '2025', '2026 YTD'],
    y=[3.6, 8.5, 5.1, 4.2, 3.9, 4.3, 4.8, 4.5],
    markers=True,
    title='Lansing-East Lansing MSA Unemployment Rate (Annual / YTD Average)',
    labels={'x': 'Year', 'y': 'Unemployment Rate (%)'}
)
fig.update_layout(template='plotly_white', height=420, title_x=0.5)
fig.show()
```

## 4. Employment Footprint Snapshot (QCEW)

QCEW provides the primary public employment series by ownership and industry. Key ownership codes:

- **0** Total
- **2** State government (includes public universities such as MSU)
- **3** Local government
- **5** Private

Industry stand-ins for the auto cluster include NAICS 336 (transportation equipment) and 3361 (motor vehicle manufacturing). Suppression is common for single-employer cells; zeros with disclosure code "N" must be treated as missing.

```python
# Illustrative QCEW ownership composition (tri-county, recent year)
ownership_data = pd.DataFrame({
    'Ownership': ['Private', 'State Government', 'Local Government', 'Federal'],
    'Employment': [155000, 32000, 18000, 2000],
    'Share (%)': [75.1, 15.5, 8.7, 1.0]
})
print('Illustrative tri-county employment by ownership (order-of-magnitude)')
display(ownership_data)

fig = px.bar(
    ownership_data,
    x='Ownership',
    y='Employment',
    text='Share (%)',
    title='Tri-County Employment by Ownership (Illustrative)',
    color='Ownership',
    color_discrete_sequence=px.colors.qualitative.Set2
)
fig.update_traces(texttemplate='%{text:.1f}%', textposition='outside')
fig.update_layout(template='plotly_white', height=420, title_x=0.5, showlegend=False)
fig.show()
```

## 5. Resident Outcomes Snapshot (ACS 2020–2024)

American Community Survey 5-year estimates provide the baseline for resident income, poverty, housing tenure, and labor force status across the three counties and principal cities.

```python
# Key ACS indicators (2020-2024 5-year, counties)
acs_key = pd.DataFrame({
    'Geography': ['Clinton County', 'Eaton County', 'Ingham County'],
    'Median Household Income': [88210, 79597, 62000],  # Ingham approx from typical pattern; replace with exact load
    'Poverty Rate (%)': [9.7, 8.3, 18.5],
    'Owner-Occupied (%)': [82.2, 74.2, 55.0]
})
print('ACS 2020–2024 Key Resident Indicators (counties)')
display(acs_key.round(1))

fig = px.bar(
    acs_key,
    x='Geography',
    y='Median Household Income',
    title='Median Household Income by County (ACS 2020–2024)',
    color='Geography',
    color_discrete_sequence=px.colors.qualitative.Pastel
)
fig.update_layout(template='plotly_white', height=400, title_x=0.5, showlegend=False)
fig.show()
```

## 6. Big 4 Context Notes

| Institution | Primary Public Indicators Available | Current Stand-ins / Caveats |
|-------------|-------------------------------------|-----------------------------|
| **MSU** | IPEDS finance & staff; QCEW own_code = 2 (state gov includes universities) | Do not double-count MSU headcount on top of state-government QCEW. Transparency reports available for cross-check. |
| **GM** | QCEW NAICS 336 / 3361 (Eaton published; Ingham suppressed); Lansing ACFR principal employers & taxpayers; parcel taxable value | Report employment as a range. Ultium/LG Energy Solution treated separately. |
| **State Government** | QCEW own_code = 2; Michigan Civil Service Commission workforce report | Statewide classified workforce as context beside local QCEW. |
| **Local / Regional Gov** | QCEW own_code = 3; Lansing ACFR; Michigan STC taxable valuations | Principal employers and taxpayers transcribed from FY2025 ACFR. |

**Commuting / retention note:** LODES (administrative residence) and ACS commuting flows differ substantially for Ingham workplaces. Report both as bounds rather than a single capture rate.
